119092 has the highest sector concentration with an HHI of 0.2064, classified as Moderate concentration. Higher HHI values indicate greater concentration in fewer sectors.

---

## Conclusion

The advanced analytics indicate that fund risk, investor
behaviour, SIP continuity and portfolio concentration should
all be considered when evaluating mutual fund portfolios.

Historical VaR and CVaR highlight downside risk, cohort
analysis identifies stronger investor groups, SIP continuity
helps identify potentially disengaging investors, and HHI
highlights funds with concentrated sector exposure.

These insights can be used alongside Sharpe ratio, Sortino
ratio, drawdown and other performance metrics for a more
complete investment analysis.
