```python
import pandas as pd

fund_master = pd.read_csv("data/raw/01_fund_master.csv")
nav_history = pd.read_csv("data/raw/02_nav_history.csv")

missing_codes = fund_master[
    ~fund_master["amfi_code"].isin(nav_history["amfi_code"])
]

print("Total AMFI codes in fund_master:", fund_master["amfi_code"].nunique())
print("Total AMFI codes in nav_history:", nav_history["amfi_code"].nunique())
print("Missing AMFI codes:", len(missing_codes))

if len(missing_codes) == 0:
    print("All AMFI codes in fund_master exist in nav_history.")
else:
    print("Missing AMFI Codes:")
    print(missing_codes[["amfi_code", "scheme_name"]])
```
